<h1 style="font-size:28px;">
    HANDLING MISSING VALUES
</h1>

</div>

  Handling missing values in the column **"unit"**

```python
 df['unit'].unique()
```

```python
df['unit'].value_counts()
```

```python
df['unit'] = df['unit'].replace(['"'],np.nan)
```

```python
df['unit'].value_counts()
```

```python
df['unit'] = df.groupby('commodity')['unit'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown')
)
```

```python
df['unit'].unique()
```

```python
# Cross-Checking
df['unit'].isnull().sum()
```

* Missing values in the unit column were handled using the mode of the corresponding commodity. 
Since a commodity is generally associated with a specific measurement unit, commodity-wise 
mode imputation provides a more meaningful replacement than using the overall most frequent unit.

```python
df['unit'] = df['unit'].replace({
    'Kg': 'Kgs',
    'Cum': 'Cbm',
    'Ton': 'Mts'
})
```

```python
df['unit'].unique()
```

* In the dataset ,the unit column contained inconsistent representations of the same unit of measurement(eg,'Kg': 'Kgs',
'Cum': 'Cbm','Ton': 'Mts').Such inconsistencies can affect grouping and analysis.To resolve this, the replace() function 
in pandas was used to standardize the values.

 Handling missiing values in the column **"value_qt"**

```python
# Group by both commodity and unit to fill values accurately based on item type and scale.
df['value_qt'] = df.groupby(['commodity', 'unit'])['value_qt'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
# Filling remaining missing values using overall median
df['value_qt'] = df['value_qt'].fillna(df['value_qt'].median())
```

```python
# Cross-Checking
df['value_qt'].isnull().sum()
```

* Missing quantities were imputed using the median quantity within the same commodity and measurement unit.
Median imputation was selected because export quantities may be highly skewed and can contain extreme values. 
Any remaining missing observations were filled using the overall median quantity.

 Handling missiing values in the column **"value_rs"**

```python
df['value_rs'] = df.groupby(['commodity', 'unit'])['value_rs'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df['value_rs'].isnull().sum()
```

* Missing export values in Indian Rupees were imputed using the median value within the corresponding commodity 
and measurement unit. This preserves differences in the scale of different commodities and reduces the influence 
of extreme export values.

 Handling Zero and Missing Values in **"value_dl"**

```python
# Calculating the actual exchange rate (INR per USD) from valid rows
# Using median to avoid any rare data entry typos in individual rows
implied_exchange_rate = (df['value_rs']/df['value_dl']).median()
# filling missing USD values by dividing the Rupees value by the exchange rate
df['value_dl'] = df['value_dl'].fillna(df['value_rs']/implied_exchange_rate)
```

```python
df['value_dl'].describe()
```

Step 1 — Check how many zeros contains

```python
# Count zero values in value_dl
zero_dl = (df['value_dl'] == 0).sum()

print("Number of zero values in value_dl:", zero_dl)
print("Percentage of zero values:",
      round(zero_dl / len(df) * 100, 2), "%")
```

Step 2 — Check whether zero USD values have non-zero INR values

```python
# Check records where USD value is zero but INR value is positive
zero_dl_positive_rs = df[
    (df['value_dl'] == 0) &
    (df['value_rs'] > 0)
]

print("Rows with value_dl = 0 but value_rs > 0:",
      len(zero_dl_positive_rs))

zero_dl_positive_rs[['value_rs', 'value_dl', 'value_qt']].head()
```

Step 3 — check cases where zero USD and zero INR together

```python
zero_both = df[
    (df['value_dl'] == 0) &
    (df['value_rs'] == 0)
]

print("Rows where both value_dl and value_rs are zero:",
      len(zero_both))
```

Step 4 — Also check quantity:

```python
zero_all = df[
    (df['value_dl'] == 0) &
    (df['value_rs'] == 0) &
    (df['value_qt'] == 0)
]

print("Rows where USD, INR and quantity are all zero:",
      len(zero_all))
```

Step 5 — Convert suspicious zero USD values to missing values

```python
df.loc[
    (df['value_dl'] == 0) & (df['value_rs'] > 0),
    'value_dl'
] = np.nan
```

```python
df['value_dl'].isnull().sum()
```

Step 6 — To fill the missing value_dl:

A better approach is to calculate a year-wise median INR/USD relationship.

```python
df['year'] = df['date'].dt.year
```

```python
# Then calculate the yearly implied exchange rate:
valid_exchange = df[
    (df['value_rs'] > 0) &
    (df['value_dl'] > 0)
].copy()

yearly_exchange_rate = (
    valid_exchange
    .assign(
        exchange_rate=lambda x: x['value_rs'] / x['value_dl']
    )
    .groupby('year')['exchange_rate']
    .median()
)

yearly_exchange_rate
```

```python
# Then fill missing USD values:
df['exchange_rate'] = df['year'].map(yearly_exchange_rate)

df['value_dl'] = df['value_dl'].fillna(
    df['value_rs'] / df['exchange_rate']
)
```

* It estimates the USD value using the typical INR/USD relationship for that particular year, rather than assuming one 
exchange rate applies to the entire dataset.


```python
df['value_dl'].isnull().sum()
```

The value_dl column represents the export value in US Dollars. During data validation, zero values were investigated 
rather than automatically treated as missing values. Records where value_dl was zero while value_rs contained 
a positive export value were identified as potentially inconsistent records. Such zero values were converted to 
missing values and subsequently imputed using a year-wise median exchange-rate relationship between INR and USD. 
    This approach helps preserve the available export-value information while reducing the effect of inconsistent zero 
entries.

    DETECTING OUTLIERS USING IQR
</h1>

</div>

```python
len(df)
```

```python
df.describe()
```

An analysis of the dataset's summary statistics (df.describe()) reveals severe, high-magnitude outliers 
primarily concentrated in the numerical trade and transaction value columns: value_qt (Quantity), value_rs (Rupee Value), 
and value_dl (Dollar Value).The outliers are characterized by a massive, disproportionate leap between the 
75th percentile (Q3) and the maximum value across all three metrics:

* value_qt (Quantity): 75% of the data points have a quantity value of 28.5 or lower, yet the maximum value reaches an astronomical 558,096,400.
    
* value_rs (Rupee Value): The 75th percentile is sits tightly at 11.03, while the maximum value spikes heavily up to 326,038.2.
    
* value_dl (Dollar Value): 75% of the dataset consists of tiny decimal values at or below 0.03, but the peak maximum recorded value reaches 445,960.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Plot each column on its own scale
df.boxplot(column='value_qt', ax=axes[0])
axes[0].set_title('Quantity Outliers')

df.boxplot(column='value_rs', ax=axes[1])
axes[1].set_title('Rupee Value Outliers')

df.boxplot(column='value_dl', ax=axes[2])
axes[2].set_title('Dollar Value Outliers')

plt.tight_layout()
plt.show()
```

```python
# Check outliers in all important numerical columns
numeric_columns = [
    'value_qt',
    'value_rs',
    'value_dl'
]

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(
        f"{col}: {len(outliers):,} outliers "
        f"({len(outliers)/len(df)*100:.2f}%)"
    )
```

```python
# create an outlier flag
Q1 = df['value_dl'].quantile(0.25)
Q3 = df['value_dl'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df['value_dl_outlier'] = np.where(
    (df['value_dl'] < lower_bound) |
    (df['value_dl'] > upper_bound),
    'Outlier',
    'Normal'
)
```

```python
# This allows to retain the original data while identifying extreme observations.
df['value_dl_outlier'].value_counts()
```

Outliers were identified using the Interquartile Range (IQR) method. Since unusually high export values may represent genuine large-scale international transactions rather than data-entry errors, the observations were not automatically deleted. An outlier flag was created to distinguish extreme observations from normal observations while preserving the original dataset. Where required for statistical analysis, 
capped values can be used to reduce the influence of extreme observations.

```python
country_pareto = (
    df.groupby("country_name")["value_dl"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

country_pareto["cumulative_percentage"] = (
    country_pareto["value_dl"].cumsum()
    / country_pareto["value_dl"].sum()
    * 100
)

top_n = 15
pareto = country_pareto.head(top_n)

fig, ax1 = plt.subplots(figsize=(14, 7))

fig.patch.set_facecolor("#1F4E78")
ax1.set_facecolor("#1F4E78")

ax1.bar(
    pareto["country_name"],
    pareto["value_dl"],
    color="#D0E1FD"
)

ax1.set_ylabel(
    "Export Value (USD)",
    color="white",
    fontsize=14,
    fontweight="bold"
)

ax1.set_xlabel(
    "Country",
    color="white",
    fontsize=14,
    fontweight="bold"
)

ax1.tick_params(axis="x", colors="white", rotation=60)
ax1.tick_params(axis="y", colors="white")

ax2 = ax1.twinx()

ax2.plot(
    pareto["country_name"],
    pareto["cumulative_percentage"],
    marker="o",
    linewidth=2,
    color="white"
)

ax2.set_ylabel(
    "Cumulative Export Share (%)",
    color="white",
    fontsize=14,
    fontweight="bold"
)

ax2.tick_params(axis="y", colors="white")
ax2.set_ylim(0, 105)

plt.title(
    "PARETO ANALYSIS OF EXPORT VALUE BY COUNTRY",
    color="white",
    fontsize=17,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

* The Pareto chart ranks African countries according to their cumulative export value and overlays the cumulative percentage contribution. It helps identify whether India's export value is concentrated among a relatively small number of African markets.
